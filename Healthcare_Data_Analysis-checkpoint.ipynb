```python
import pandas as pd
df = pd.read_csv('healthcare_dataset.csv')
df.head()
```

```python
print("Total Rows & Columns:", df.shape)
print("\nMissing Values Count:")
print(df.isnull().sum())
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Medical Condition', hue='Medical Condition', palette='Set2', legend=False)
plt.title('Patient Count by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()


plt.figure(figsize=(6, 4))
df['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['skyblue', 'lightpink']
)
plt.title('Gender Distribution')
plt.ylabel('')
plt.show()


plt.figure(figsize=(10, 5))
df.groupby('Medical Condition')['Billing Amount'].mean().plot(
    kind='bar',
    color='teal'
)
plt.title('Average Billing Amount by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Billing Amount ($)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Patient count per condition
print("--- Patient Count per Condition ---")
print(df['Medical Condition'].value_counts())

# Average billing amount per condition
print("\n--- Average Billing Amount per Condition ---")
print(df.groupby('Medical Condition')['Billing Amount'].mean().round(2))

# Top 5 hospitals by patient count
print("\n--- Top 5 Hospitals by Patient Count ---")
print(df['Hospital'].value_counts().head(5))

# Insurance provider distribution
print("\n--- Insurance Provider Breakdown ---")
print(df['Insurance Provider'].value_counts())
```
